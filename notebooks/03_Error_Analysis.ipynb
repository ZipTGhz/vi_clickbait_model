# 03. So sánh Mô hình, Giải thích Đặc trưng & Phân tích Lỗi (Model Comparison, Explainability & Error Analysis)
Notebook này tổng hợp kết quả thực nghiệm của cả 3 nhóm mô hình (**Baseline ML**, **Deep Learning**, **Transformer SOTA - PhoBERT-base-v2**) và đi sâu vào phân tích lỗi:
1. **Bảng so sánh toàn diện các mô hình trên tập Test** (Accuracy, Precision, Recall, F1 Clickbait, Macro F1, ROC-AUC, Thời gian huấn luyện & suy diễn).
2. **Giải thích mô hình (Feature Importance)**: Top 20 n-grams có trọng số dương lớn nhất (`clickbait`) và Top 20 n-grams có trọng số âm lớn nhất (`non-clickbait`).
3. **Phân tích lỗi chuyên sâu (Error Analysis) của `PhoBERT-base-v2`**: Khảo sát 10 mẫu **False Positive** và 10 mẫu **False Negative** dưới góc độ ngôn ngữ học.

```python
import json
import pandas as pd
from IPython.display import Image, display

pd.set_option('display.max_colwidth', None)
comp_df = pd.read_csv('../outputs/results/model_comparison.csv', encoding='utf-8')
display(comp_df.sort_values(by='macro_f1', ascending=False))
```

## 2. Giải thích Mô hình: Top 20 Từ khóa / N-grams Đặc trưng cho `Clickbait` vs `Non-Clickbait`

```python
with open('../outputs/results/feature_importance.json', 'r', encoding='utf-8') as f:
    feat_imp = json.load(f)

print('=== Top 20 N-grams hướng về nhãn CLICKBAIT (+) ===')
display(pd.DataFrame(feat_imp['logistic_regression']['top_20_clickbait_ngrams']))
print('=== Top 20 N-grams hướng về nhãn NON-CLICKBAIT (-) ===')
display(pd.DataFrame(feat_imp['logistic_regression']['top_20_non_clickbait_ngrams']))
display(Image(filename='../outputs/figures/09_feature_importance_top20.png'))
```

## 3. Phân tích Lỗi của Mô hình SOTA (`PhoBERT-base-v2`): 10 False Positives & 10 False Negatives

```python
with open('../outputs/results/phobert_error_analysis.json', 'r', encoding='utf-8') as f:
    err_data = json.load(f)

print(f"Tổng số mẫu Test: {err_data['total_test_samples']} | Tổng số lỗi: {err_data['total_errors']} (FP={err_data['total_false_positives']}, FN={err_data['total_false_negatives']})")
print('\n=== 10 MẪU FALSE POSITIVE (Tin chính thống bị đoán nhầm là Clickbait) ===')
display(pd.DataFrame(err_data['false_positives_top10']))

print('\n=== 10 MẪU FALSE NEGATIVE (Tin Clickbait tinh vi bị mô hình bỏ sót) ===')
display(pd.DataFrame(err_data['false_negatives_top10']))
```
